```python
def smart_home():
    temp=float(input("Enter Temperature : "))
    light=float(input("Enter Light Intensity Level : "))
    person=input("Is Person is there inside room : ")
    if(person=="yes"):
        if(temp>35):
            print("AC : ON")
        else:
            print("AC : OFF")
        if(light<30):
            print("Light : ON")
        else:
            print("Light : OFF")
    else:
        print("No one is there inside room ")
        print("Light : OFF")
        print("AC : OFF")
while True:
    print("----------Smart Home Intelligent System-----------")
    print("1. Smart Home")
    print("2. Exit")
    choice=int(input("Enter your Choice : "))
    if(choice==1):
        smart_home()
    elif(choice==2):
        print("Exiting....")
        break
    else:
        print("Invalid Operation")
```
